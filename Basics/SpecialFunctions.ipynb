## range
---
# Syntax range(start_value, stop_value, increment_value)

```python
a = range(10)       # generate range from 0 to number before 10
print(a, list(a))

b = range(2,10)     # generate range from 2 to number before 10
print(b, list(b))

c = range(2,10,3)   # generate range from 2 to number before 10 increment by 3
print(c, list(c))

d = range(2, -10, -2)
print(d, list(d))
```

## map
---
# arg1: function arg2: iterable op: apply arg1 to iterable in arg2

```python
my_list : list[str] = ["I", "drove", "very", "fast"]
length = map(len, my_list)  
print(list(length))

cust=map(lambda x: x + 's', my_list) # Lamba function add 's' to all objects in list
print(list(cust))

def add_es (inp : str) -> str:
    return inp + "es"

cust2=map(add_es, my_list)
print(list(cust2))
```

# filter
---
### Take all items from iterable object and apply compatible function

```python
my_list : list[str] = ["I", "drove", "very", "fast"]

op = filter(lambda x : len(x) > 4, my_list )
print(list(op))

def long_than_3(inp: str) -> bool:
    return len(inp) > 3

op2 = filter(long_than_3, my_list )
print(list(op2))
```

```python
inp = {1, 2, 3, 4, 5}

print(sum(inp))
print(sum(inp, start=100))
print(sum(inp, start=-10))


inp = (4,6,3,7,9,2,0)
print(sorted(inp))
print(sorted(inp, reverse=True))


people = [
    {"name": "ABC", "age": 26},
    {"name": "DEF", "age": 16},
    {"name": "SRT", "age": 56},
    {"name": "XYZ", "age": 76},
]

print(sorted(people, key= lambda people : people["age"]))   # lambda fn to sort based on age
print(sorted(people, key= lambda people : people["age"], reverse=True))
```

# zip
---
### combine different iterable objects together and end when one runs out

```python
names= ["ABC", "DEF", "IJK", "XYZ"]
age = [12, 18, 36, 48, 81]

combined = list(zip(names, age))
print(combined, type(combined))

for name, age in combined:
    print(f"{name} is {age} years old")
```

# Decorators

----
Decorators are function that modify other functions

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()            # Start Time
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"Function {func.__name__} took {(end_time - start_time):.4} sec to complete")
        return result
    return wrapper

@timer
def sample_fn(n : int) -> int:
    return sum(range(0, n+1))

print(sample_fn(1000000))
```
