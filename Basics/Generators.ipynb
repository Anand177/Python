# Generator

```python
def gen_values():
    yield 1
    yield 2
    yield 3

print(gen_values()) # Method isn't invoked onlyan object is returned
```

```python
gen_values_obj= gen_values()

print(next(gen_values_obj))
print(next(gen_values_obj))
print(next(gen_values_obj))
print(next(gen_values_obj))
```

# Invoke Generator using markdown

```python
for val in gen_values():
    print(val)
```

```python
def gen_prime(start: int, end: int):
    """Use Generator to get prime numbers between range of numbers provided"""
    is_prime : bool = False
    for num in range (start, end+1):
        if num < 2:
            continue
        is_prime = True
        for i in range (2, num):
            if num %i == 0:
                is_prime = False
                break
        if is_prime:
            yield num

gen_prime_obj = gen_prime(50, 100)
print(next(gen_prime_obj))
print(next(gen_prime_obj))
print(next(gen_prime_obj))
```
